```python
import requests
from bs4 import BeautifulSoup

url = "https://kur.doviz.com/"

html = requests.get(url).text
soup = BeautifulSoup(html, "html.parser")

print(soup.title)
```

```python
for etiket in soup.find_all(["a", "div", "span"]):
    yazi = etiket.get_text(" ", strip=True)
    if "Dolar" in yazi or "Euro" in yazi:
        print(yazi)
```

```python
for satir in soup.find_all("tr"):
    veriler = [x.get_text(strip=True) for x in satir.find_all(["th", "td"])]

    if veriler:
        print(veriler)
```

```python
for satir in soup.find_all("tr"):
    veriler = [x.get_text(" ", strip=True) for x in satir.find_all(["th", "td"])]

    if len(veriler) >= 6:
        print(veriler)
```

```python
for satir in soup.find_all("tr"):
    veriler = [x.get_text(" ", strip=True) for x in satir.find_all(["th", "td"])]

    if len(veriler) >= 6 and veriler[0] != "":
        print(
            "Döviz:", veriler[0],
            "| Alış:", veriler[1],
            "| Satış:", veriler[2],
            "| Yüksek:", veriler[3],
            "| Düşük:", veriler[4],
            "| Değişim:", veriler[5]
        )
```
